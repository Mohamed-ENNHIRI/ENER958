## II. AC power estimation

Modeling PV power from ghi / dhi / dni goes into four steps:

1. Modeling the POA irradiance.
2. Modeling the module cell temperature.
3. Modeling the DC power.
4. Modeling the inverter efficiency and the AC power.

### II.1 POA irradiance

#### Exercice

Calculate the POA irradiance with the isotropic assumption, as introduced in the notebook `python_intro_poa`, and compare with `actual["g_poa"]` over one week.

```python
beta = 30
azimuth_inst = 180
rho = 0.2

# The lazy way with the pvlib package
from pvlib.irradiance import get_total_irradiance
poa_data = get_total_irradiance(beta, azimuth_inst, solar_position["zenith"], solar_position["azimuth"], weather_data["dni"], weather_data["ghi"], weather_data["dhi"], albedo=rho)

fig, ax = plt.subplots(figsize=(9, 4))
poa_data.loc["2023-07-01":"2023-07-07", "poa_global"].plot(ax=ax, label="Estimated, isotropic")
actual.loc["2023-07-01":"2023-07-07", "g_poa"].plot(ax=ax, label="Actual", linestyle="--")
plt.ylabel("POA irradiance [W/m²]")
plt.legend()
```

### II.2 Cell temperature model

#### Exercice

Calculate the module temperature with, both, the NOCT and the Faiman model, and compare with `actual["t_mod"]` over four days.

```python
NOCT = 45 # From the datasheet

# Generic coefficients for an open-rack PV installation
u_0 = 25.0 # W/(°C.m2)
u_1 = 6.84 # W.s/(°C.m3)

def noct_model(ta, poa):
    k_ross = (NOCT - 20) / 800
    return ta + k_ross * poa

def faiman_model(ta, poa, ws):
    return ta + poa / (u_0 + u_1 * ws)

t_mod = noct_model(weather_data["temp_air"], poa_data["poa_global"])
t_mod2 = faiman_model(weather_data["temp_air"], poa_data["poa_global"], weather_data["wind_speed"])

fig, ax = plt.subplots(figsize=(9, 4))
t_mod.loc["2023-07-01":"2023-07-04"].plot(ax=ax, label="NOCT model")
t_mod2.loc["2023-07-01":"2023-07-04"].plot(ax=ax, label="Faiman model")
actual.loc["2023-07-01":"2023-07-04", "t_mod"].plot(ax=ax, label="Actual", linestyle="--")
plt.ylabel("Module temperature [°C]")
plt.legend()
```

### II.3 DC power model

#### Exercice

Calculate the DC power with the PVWatts model (write a function) with both temperature models.

```python
Pdc_0 = 3000 # Wp, nominal power
gamma_pdc = -0.35 / 100 # [-]/°C

def pvwatts_dc(poa_global, t_mod, gamma_pdc):
    dc_power = Pdc_0 * poa_global / 1000 * (1 + gamma_pdc * (t_mod - 25))
    return dc_power

dc_power = pvwatts_dc(poa_data["poa_global"], t_mod, gamma_pdc)
dc_power2 = pvwatts_dc(poa_data["poa_global"], t_mod2, gamma_pdc)

fig, ax = plt.subplots(figsize=(9, 4))
dc_power.loc["2023-07-01":"2023-07-04"].plot(ax=ax, label="DC power (NOCT temp. model)")
dc_power2.loc["2023-07-01":"2023-07-04"].plot(ax=ax, label="DC power (Faiman temp. model)")
actual.loc["2023-07-01":"2023-07-04", "p_dc"].plot(ax=ax, label="Actual", linestyle="--")
plt.ylabel("DC power [W]")
plt.legend() # Print the labels
```

### II.4 AC power model

The inverter efficiency depends on the load. The PVWatts inverter model:

$\eta = \frac{\eta_{nom}}{\eta_{ref}} (-0.0162 \zeta - \frac{0.0059}{\zeta} + 0.9858)$ with $\zeta = P_{DC} / P_{DC,0}$, $\eta_{nom}$ = 0.96 and $\eta_{ref}$ = 0.9637.

The AC power is $P_{AC} = \eta \cdot P_{DC}$, capped at the AC rating of the inverter $P_{AC,0}$.

#### Exercice

Calculate the AC power with the PVWatts inverter model (write a function). Then compare the estimated AC power with the actual AC power over one week.

```python
Pac_0 = 2500 # W, AC rating of the inverter

# Adapted from pvlib
def pvwatts_inverter(pdc, pdc0, pac0, eta_inv_nom=0.96, eta_inv_ref=0.9637):
    zeta = pdc / pdc0
    zeta = zeta.replace(0, np.nan) # Avoid to divide by 0
    eta = eta_inv_nom / eta_inv_ref * (-0.0162 * zeta - 0.0059 / zeta + 0.9858)
    eta = eta.clip(lower=0) # No negative efficiency at very low load
    pac = (eta * pdc).fillna(0)
    pac = pac.clip(upper=pac0) # Inverter clipping
    return pac

ac_power = pvwatts_inverter(dc_power2, Pdc_0, Pac_0)

fig, ax = plt.subplots(figsize=(9, 4))
ac_power.loc["2023-07-01":"2023-07-07"].plot(ax=ax, label="Estimated AC power")
actual.loc["2023-07-01":"2023-07-07", "p_ac"].plot(ax=ax, label="Actual AC power", linestyle="--")
plt.ylabel("AC power [W]")
plt.legend()
```

## III. Performance metrics

In this section, the performance of the installation and the accuracy of the models are assessed.

- Performance ratio: $PR = \frac{E_{AC}}{P_0 \cdot H_{POA} / 1000}$, with $H_{POA}$ the annual POA irradiation in Wh/m²
- Energy performance index: $EPI = \frac{E_{AC, actual}}{E_{AC, estimated}}$

#### Exercice

Calculate the PR of the actual system and the PR expected from the model, then the EPI.

```python
PR = actual["p_ac"].sum() / (Pdc_0 * actual["g_poa"].sum() / 1000)
print(f"PR actual: {round(PR * 100, 1)} %")

# From the estimation in section II
PR_mod = ac_power.sum() / (Pdc_0 * poa_data["poa_global"].sum() / 1000)
print(f"PR expected: {round(PR_mod * 100, 1)} %")

EPI = actual["p_ac"].sum() / ac_power.sum()
print(f"EPI: {round(EPI * 100, 1)} %")
```

#### Exercice, to go further

What is the accuracy of the estimated values compared to the actual ones ? Calculate the mean bias error (MBE), the mean absolute error (MAE) and the root mean square error (RMSE) of the POA irradiance, the module temperature and the AC power, for the daytime hours (g_poa > 10 W/m²). Which variable carries the largest relative error?

```python
def errors(estimated, measured, unit):
    error = estimated - measured
    print(f"MBE: {round(error.mean(), 2)} {unit} | MAE: {round(error.abs().mean(), 2)} {unit} | RMSE: {round((error ** 2).mean() ** 0.5, 2)} {unit} | nRMSE: {round(100 * (error ** 2).mean() ** 0.5 / measured.mean(), 1)} % of the mean")

filter = actual["g_poa"] > 10 # Daytime
print("POA irradiance"); errors(poa_data.loc[filter, "poa_global"], actual.loc[filter, "g_poa"], "W/m²")
print("Module temperature"); errors(t_mod2.loc[filter], actual.loc[filter, "t_mod"], "°C")
print("AC power"); errors(ac_power.loc[filter], actual.loc[filter, "p_ac"], "W")
```
